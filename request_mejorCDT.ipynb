```python
import requests
#requests permite enviar solicitudes HTTP.
#En este caso, se utiliza para hacer una solicitud GET a la URL


#ESTA FUNCIÓN RECIBE COMO ARGUMENTO UNA URL
def check_website_status(url):

#Se hace uso de try except para que el programa no pare en caso tal de que se tenga algún error

    try:
        #Se hace una solicitud GET a la URL
        response = requests.get(url)
        #Se verifica el estado de la respuesta
        if response.status_code == 200:
            #Si el estado es 200, se muestra un mensaje de que la pagina está habilitada
            print(f"Website {url} is up and running.")
        else:
            #Si el estado es diferente de 200, se muestra un mensaje de que la pagina no está habilitada
            print(f"Website {url} is down. Status code: {response.status_code}")

    #Si hay un error en la solicitud, se muestra un mensaje de error
    except requests.exceptions.RequestException as e:
        print(f"Error checking {url}: {e}")

if __name__ == "__main__":
    check_website_status("https://asj.com")
```
